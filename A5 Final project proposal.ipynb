## Overview & Research Questions

Going back to that all-team meeting: “there’s many ways to know but also we don’t know” is not the answer I want to give when someone asks me the same question one day. So the goal of this project is to replace a little bit of that “we don’t know” with actual data.

Specifically, I want to figure out whether when a company goes public has any relationship with how well it does afterward. “When” can mean a few different things, so I’m splitting it into four questions:

Company age: How does company age at IPO affect long-term success?
Time of year: How does the month or season of the IPO affect their long-term performance?
Election cycle: How do companies that go public in a presidential election year (or the year right after) perform differently from companies that go public mid-cycle?
Market conditions: How do companies that go public during “hot” IPO years (when lots of other companies are also going public) differ from companies that go public in quieter years?

“Long-term success” here means how the stock does 1, 3 and 5 years after the IPO, compared to the overall market (the S&P 500). A company whose stock went up 40% isn’t really a success if the whole market went up 80% in that same time.

### My hypotheses, based on what I found in the research (more on that below):

H1: Older companies will have better long-term returns than younger ones.
H2: IPO month won’t matter much. I’m expecting this to be a “no real effect” result, which is still a useful answer.
H3: Election-year IPOs will be less common, and their long-term returns will differ from mid-cycle IPOs. I’m not sure which direction yet.
H4: Companies that go public in hot IPO years will have worse long-term returns. This one is the most counterintuitive to me, because it means the moment everyone is excited might be the worst moment to jump in.

## Background / Related Work

It turns out I’m not the first person to wonder about this, and the existing research shaped this project a lot.

**Company age**. A study of 1,234 IPOs from 1991–1997 found a statistically significant positive relationship between a company’s age at IPO and how its stock performed afterward (Clark, 2001). A more recent study of 1,611 IPOs from 2009–2019 found that both age and size help predict future stock performance, and that tech and healthcare companies tend to go public younger than other sectors (Siev & Qadan, 2022). That second point matters for me, since it means I can’t just compare all companies directly without accounting for industry. A smaller study of internet companies found a U-shaped pattern instead: the very young and the very old companies did best (Strottner, 2017). So the age question isn’t fully settled, which makes it worth testing on a bigger, longer dataset. It’s also very relevant right now, because companies are waiting longer than ever. Average age at IPO was 9.5 years from 1980–2019 but hit 14 years in 2024 (Ritter, via The Corporate Counsel, 2025).

**Elections**. Researchers studying U.S. governor elections found fewer IPOs from a state when it has an election coming up. Companies that did go public in those years got lower offer prices, meaning raising money cost them more (Çolak, Durnev & Qian, 2017). That worked out to about 15% fewer IPOs in election years, with a rebound the year after. On the other hand, PwC’s look at presidential elections says IPO volume is mostly consistent across election and non-election years, apart from a dip in November (PwC, 2024). Those two findings don’t totally agree, which is exactly why I want to look at it myself.

**Market timing**. This was the most interesting thing I found. A 2026 study, “Anomalies and the Decision to Go Public,” found that private companies tend to go public when similar public companies are temporarily overpriced. Those companies go public almost two years sooner, and then underperform by about 20% over the next five years (Horstman, Li & Zhong, 2026, summarized by Swedroe). Basically, going public when the market is hot is great for raising money but not necessarily great for long-term success. That’s where H4 comes from.

**Time of year**. I couldn’t find much serious research on whether IPO month affects long-term performance. Most of what’s out there is about deal volume (when companies choose to go public), not about results. That’s a gap my project can actually fill, even if the answer ends up being “it doesn’t matter.”

Overall, the research told me three things: (1) I need a big dataset of many companies, not just five famous ones; (2) I need to measure returns relative to the market, not raw stock price; and (3) industry is a confounding factor I need to control for.
