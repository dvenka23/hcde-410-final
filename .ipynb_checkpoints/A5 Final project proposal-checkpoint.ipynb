# A5: Final Project Proposal
### How does the time and starting valuation of a company affect its longterm success and valuation?
## Motivation & problem
I've had the opportunity to work in and be completely surrounded by startups and the startup life throughout my life. My parents have often opted to work in smaller, private companies which have then gone public. In fact, I just worked in a Series A startup over the summer and remember sitting in an all-team meeting discussing the timeline for going public. A coworker of mine asked how the company would decide to do that and the leadership team essentially said "there's many ways to know but also we don't know."

Not a super helpful answer, to say the least.

I, of course, want this company to be successful so that I can have a big name on my resume. Unfortunately, the market is tough to understand, but I'd like to know if there's any correlation between when a company goes public (in terms of time since the company started, the time of the year, the year in relation to external factors like time since the last election, and more). Hopefully this information could be used to inform when this company (and hopefully manner others in my future) go public.

## Data
I will be using this dataset:
**Founding dates for firms going public in the U.S. during 1975-2025**-https://site.warrington.ufl.edu/ritter/files/founding-dates.pdf 

This dataset includes data about multiple companies's IPO dates and founding dates alongside some other general information about the company itself. This data alongside the yfinance library (yahoo finance library) in Python will help me correlate the company against the stock value per year to show growth/decline rates.

This information is really useful because it will allow me to track the patterns of each of the companies' stocks to see which is the most stable, which has the most growth, and which is most desired, and then compare this information against the time at which they were made public. Of course, this is a difficult thing to compare equally because of the choices the company themselves may make about their stocks (e.g. splitting stocks), but this will be considered by analyzing against the total rather than individual valuations (alongside the individual approach). There are also social considerations to be made. Essentially, I cannot compare these stock against each other without including the social factors that could impact their trends--like Facebook's data controversies, Netflix adding ads, Google promising certain sustainability measure, etc. And abnormalities can be cross-referenced against social factors when noticed.

## Unknowns & dependencies
Due to the extremely public nature of all these companies, there's a chance I find out that they actually have too many scandals and normalizing the data could take too long. I currently do not know how many social factors I will need to keep track of but as long as it's not the norm and truly are outliers, I think I shoudl be fine! Moreover, I won't actually know about companies that have gone out of business which could impact the analysis at first, so I will need to make sure I exclude any companies as they go out of business or make required adjustments for name/ticker changes.
