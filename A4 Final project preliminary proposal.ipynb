# A4: Final Project Preliminary Proposal
### How does the time and starting valuation of a company affect its longterm success and valuation?
## Motivation & problem
I've had the opportunity to work in and be completely surrounded by startups and the startup life throughout my life. My parents have often opted to work in smaller, private companies which have then gone public. In fact, I just worked in a Series A startup over the summer and remember sitting in an all-team meeting discussing the timeline for going public. A coworker of mine asked how the company would decide to do that and the leadership team essentially said "there's many ways to know but also we don't know."

Not a super helpful answer, to say the least.

I, of course, want this company to be successful so that I can have a big name on my resume. Unfortunately, the market is tough to understand, but I'd like to know if there's any correlation between when a company goes public (in terms of time since the company started, the time of the year, the year in relation to external factors like time since the last election, and more). Hopefully this information could be used to inform when this company (and hopefully manner others in my future) go public.
