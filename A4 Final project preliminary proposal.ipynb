## Data
I will be using this dataset:
FAANG- Complete Stock Data--https://www.kaggle.com/datasets/aayushmishra1512/faang-complete-stock-data/data

This dataset includes data about the FAANG companies’ (Facebook, Amazon, Apple, Netflix, and Google) stocks. Specifically, it shows daily data on the opening price, closing price, high, low, adjusted closing price, and volume of trades for that day.

This information is really useful because it will allow me to track the patterns of each of the FAANG stocks to see which is the most stable, which has the most growth, and which is most desired, and then compare this information against the time at which they were made public. Of course, this is a difficult thing to compare equally because of the choice sthe company themselves may make about their stocks (e.g. splitting stocks), but this will be considered by analyzing against the total rather than individual valuations (alongside the individual approach). There are also social considerations to be made. Essentially, I cannot compare these stock against each other without including the social factors that could impact their trends--like Facebook's data controversies, Netflix adding ads, Google promising certain sustainability measure, etc. And abnormalities can be cross-referenced against social factors when noticed.
